## Split and Merge Image Chnnels

```python
import cv2
import numpy as np
img = cv2.imread('image01a.jpg')
b,g,r = cv2.split(img)
```

```python
cv2.imshow("b_image",b)
cv2.waitKey(0) 
```

```python
cv2.imshow("g_image",g)
cv2.waitKey(0)
```

```python
cv2.imshow("r_image",r)
cv2.waitKey(0) 
cv2.destroyAllWindows()
```

```python
imag=cv2.merge((b,r,g ))
```

```python
cv2.imshow("merged_image",imag)
cv2.waitKey(0)
```

```python
# Sum Images 
# Sum the two image arrays for all channels
# Read in the two images

image_1 = cv2.imread('image02a.jpg')
image_2 = cv2.imread('image02b.jpg')

result = cv2.add(image_1, image_2)

cv2.imshow('result', result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Sum Images 
image_1 = cv2.imread('image02a.jpg')
image_2 = cv2.imread('image02b.jpg')

result = cv2.addWeighted(image_1, 0.1, image_2, 0.9, 0)

cv2.imshow('result', result)
cv2.waitKey(0) # Wait for the user to press a key before continuing
cv2.destroyAllWindows()
```

### 6.  Filtering Images ###

```python
# Filter by passing image through 3x3 averaging filter
# average_image 
new_image= cv2.blur(image_1,(3,3))
```

```python
# Apply 3x3 gaussian filter on the original image
# gaussian_image
new_image = cv2.GaussianBlur(image_1,(3,3),0)
```

```python
# Apply 3x3 median filter on the original image
# median_image
new_image = cv2.medianBlur(image_1,3)
```

```python
import matplotlib.pyplot as plt

plt.imshow(new_image)
plt.show()
```

### Read from URL

```python
import cv2
import urllib.request as req 
import numpy as np
import matplotlib.pyplot as plt

req = req.urlopen('https://upload.wikimedia.org/wikipedia/commons/7/78/Pisa_-_veduta_aerea_2.JPG')
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
image_1 = cv2.imdecode(arr, -1) 

plt.imshow(image_1)
plt.show()
```
